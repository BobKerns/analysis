# Lifespans

How long each of us may live, as the plan sees it. Nobody knows their own age at death, so the plan doesn't use a single age: it uses a range of ages, each with its chance, from the Social Security Administration's 2023 life table for each sex. Only persons with a sex in `persons.csv` are shown.

Each person's curves are centered on the age at death the scenario sets in `lifespan.csv`: the table is read as if the person were a few years older or younger, so that its middle age (the median) is that age. For a `life_table` lifespan these are the curves an ensemble draws its members' ages from; for a fixed age they show the uncertainty that age stands in for. The plain table's own median is marked as a dashed orange line for comparison. Someone with no age at death set is shown with the plain table.

Everything starts at the scenario's first month: the chances are for someone alive then.

The same chart from the terminal: `analysis sim person show [PERSON] --graph`.

## Reading the charts

Each person gets a row of two charts. Every chart runs from the scenario's start to its end, the years the plan covers, and both people's "chance of dying" charts share one scale, so their bars compare directly.

- **Left, the chance of dying at each age.** Each bar is one year of age. Tall bars are the likeliest ages; the bars thin out at both ends. Their heights add up to 100%.
- **Right, the chance of having died by each age.** The solid line climbs from 0% to 100%; the dashed green line is its mirror, the chance of still being alive. Where the solid line crosses 50% is the median: half the chance falls before it, half after.

The blue shading is the same as the ensemble's fan charts. The palest band covers the middle 90% of the chances (from the 5th to the 95th percentile), the next the middle 60%, and the darkest the middle 20%. Ages outside the palest band are possible but unusual: one in twenty falls below it, one in twenty above.

The last row is the household. A plan has to last until the *last* of us dies, which is usually later than either one's own median age. The thick line is the chance that at least one of us is still alive in each year, the dotted line the chance that both are, and the shading is the fan of the year of the last death. The two lifespans are treated as independent.

## The numbers behind the charts

The tables above the charts give each person's age at death by percentile: `p05` is the age by which 5% of the chance has passed, `p50` the median, `p95` the age by which 95% has. The household row gives the same for the calendar year of the last death.

To change a lifespan: `analysis sim person lifespan set PERSON AGE --vary life_table`.
